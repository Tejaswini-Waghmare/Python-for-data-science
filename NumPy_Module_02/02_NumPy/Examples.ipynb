```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 1. Arrays and properties

```python
a = np.array([10, 20, 30])
m = np.array([[1, 2, 3], [4, 5, 6]])
print("1D:", a.shape, a.ndim, a.dtype)
print("2D:", m.shape, m.ndim, m.size)
```

## 2. Constructors

```python
print(np.zeros((2, 3)))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
```

## 3. Indexing and filtering

```python
sales = np.array([120, 450, 80, 700, 250])
print("Slice:", sales[1:4])
print(">=250:", sales[sales >= 250])
```

## 4. Views versus copies

```python
original = np.array([1, 2, 3, 4])
view = original[1:3]
view[0] = 99
print("After view change:", original)
copy = original[1:3].copy()
copy[0] = 0
print("After copy change:", original)
```

## 5. Reshaping and axes

```python
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)
print("Row sums:", matrix.sum(axis=1))
print("Column means:", matrix.mean(axis=0))
```

## 6. Vectorization and broadcasting

```python
prices = np.array([100., 200., 300.])
print("With tax:", prices * 1.08)
monthly = np.array([[10, 20, 30], [40, 50, 60]])
print("Weighted:\n", monthly * np.array([1, 2, 3]))
```

## 7. Missing values and statistics

```python
scores = np.array([72., 85., np.nan, 91.])
print("Mean ignoring missing:", np.nanmean(scores))
print("Median ignoring missing:", np.nanmedian(scores))
```

## 8. Reproducible simulation

```python
rng = np.random.default_rng(42)
sample = rng.normal(50, 10, 100)
print("Sample mean:", round(sample.mean(), 2))
print("Sample SD:", round(sample.std(ddof=1), 2))
```

## 9. Linear algebra

```python
A = np.array([[2., 1.], [1., -1.]])
b = np.array([5., 1.])
print("Solution [x,y]:", np.linalg.solve(A, b))
```

## 10. Mini project: transaction analytics

```python
amounts = np.array([120., -40., 350., -25., 90., 500.])
credits = amounts[amounts > 0]
debits = amounts[amounts < 0]
summary = {"net": amounts.sum(), "credits": credits.sum(), "debits": debits.sum(), "large_credits": np.sum(credits >= 300)}
print(summary)
assert summary["net"] == 995.0
assert summary["large_credits"] == 2
```
