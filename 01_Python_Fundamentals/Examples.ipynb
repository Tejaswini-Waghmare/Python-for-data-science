## 2. Lists and slicing

Select, filter, and summarize values.


```python
amounts = [120, -20, 80, 250, -40]
print("First three:", amounts[:3])
print("Net:", sum(amounts))
print("Credits:", [x for x in amounts if x > 0])
```

## 4. Conditions and loops

Classify transaction values.


```python
for amount in [100, -30, 0]:
    if amount > 0:
        label = "credit"
    elif amount < 0:
        label = "debit"
    else:
        label = "zero"
    print(amount, label)
```

## 5. Functions and validation

Write reusable, safe summaries.


```python
def summarize(values):
    if not values:
        raise ValueError("values must not be empty")
    return {"count": len(values), "net": sum(values), "average": sum(values) / len(values)}

print(summarize([100, -30, 60]))
```

## 6. Comprehensions and generators

Filter and transform without modifying the source.


```python
numbers = [1, 2, 3, 4, 5]
squares_of_even = [n*n for n in numbers if n % 2 == 0]
print(squares_of_even)
print(sum(n*n for n in numbers))
```

## 7. Exceptions

Handle invalid user inputs explicitly.


```python
def parse_amount(raw):
    try:
        return float(raw)
    except (TypeError, ValueError) as exc:
        raise ValueError("Amount must be numeric") from exc

for raw in ["12.5", "bad"]:
    try:
        print(parse_amount(raw))
    except ValueError as exc:
        print("Error:", exc)
```

## 9. Mini case study: expense analysis

Summarize transactions by category using pure Python.


```python
records = [
    {"category": "food", "amount": 25},
    {"category": "travel", "amount": 40},
    {"category": "food", "amount": 15},
]
totals = {}
for row in records:
    category = row["category"]
    totals[category] = totals.get(category, 0) + row["amount"]
print("Totals by category:", totals)
```
